# NBA Data Cleaning

## NBA Franchise Value Data

```python
# Load in NBA franchise value csv
data = '../data/raw/MVP2/nba_franchise_values.csv'
nba_franchise_data = pd.read_csv(data)

# create DataFrame from data
nba_franchise_df = pd.DataFrame(nba_franchise_data)
# Show DF's first 5 rows
nba_franchise_df.head()
```

```python
# Review full DataFrame
nba_franchise_df
```

```python
# Check info on DataFrame
nba_franchise_df.info()
```

```python
# Replace commas in franchise column with periods
nba_franchise_df['Average NBA Franchise Value (USD Millions)'] = nba_franchise_df['Average NBA Franchise Value (USD Millions)'].str.replace(',','.')

nba_franchise_df
```

```python
nba_franchise_df.info()
```

```python
# Add decimal to far left of 2001-2014 values to show as parts of a billion (to match format in end)
def move_period_to_left(s):
    s = s.replace('.', '')  # Remove existing period
    return '.' + s  # Add period to the far left, "period + s(represents value)"

nba_franchise_df.loc[nba_franchise_df.index >= 9, 'Average NBA Franchise Value (USD Millions)'] = nba_franchise_df.loc[nba_franchise_df.index >= 9, 'Average NBA Franchise Value (USD Millions)'].apply(move_period_to_left)

nba_franchise_df
```

```python
# Function to remove incorrect second period (decimal)
def remove_period(s):
    parts = s.split('.')
    return '.'.join(parts[:2]) + ''.join(parts[2:])

nba_franchise_df['Average NBA Franchise Value (USD Millions)'] = nba_franchise_df['Average NBA Franchise Value (USD Millions)'].apply(remove_period)

nba_franchise_df
```

```python
# Convert franchise values to float then round to 1 decimal place
# 1) change type
nba_franchise_df['Average NBA Franchise Value (USD Millions)'] = nba_franchise_df['Average NBA Franchise Value (USD Millions)'].astype(float) 

# 2) round numbers over 1 billion to 1 place after decimal
nba_franchise_df.loc[nba_franchise_df.index < 9,'Average NBA Franchise Value (USD Millions)'] = nba_franchise_df.loc[nba_franchise_df.index < 9,'Average NBA Franchise Value (USD Millions)'].round(1)

# 3) round numbers less than 1 billion to 2 places after decimal
nba_franchise_df.loc[nba_franchise_df.index >= 9,'Average NBA Franchise Value (USD Millions)'] = nba_franchise_df.loc[nba_franchise_df.index >= 9,'Average NBA Franchise Value (USD Millions)'].round(2)

nba_franchise_df
```

```python
# Rename francshise column to billions format to match previously clean DF's
nba_franchise_df = nba_franchise_df.rename(columns={'Average NBA Franchise Value (USD Millions)':'Avg NBA Franchise Value (USD billions)' })

# Review DF info after cleaning
print(nba_franchise_df.info())
nba_franchise_df
```

## NBA Revenue Data


```python
# Read in NBA revenue data and create DataFrame
nba_rev_path = '../data/raw/MVP2/nba_revenue.csv'
nba_rev_data = pd.read_csv(nba_rev_path)

# Create DataFrame
nba_rev_df = pd.DataFrame(nba_rev_data)
nba_rev_df
```

```python
# Review info behind DataFrame
nba_rev_df.info()
```

**Cleaning Notes**
- Year will likely need to be reformatted from YY/YY --> YYYY to create universal Year column for joining
- Rev column: commas to period and period added from 2016-2009
- Change value to billions

```python
# Replace commas with period (decimals)
nba_rev_df['NBA Sponsorship Revenue (USD Millions)'] = nba_rev_df['NBA Sponsorship Revenue (USD Millions)'].str.replace(',', '.')

# Check 
nba_rev_df
```

```python
# Add decimal point (period since type is still 'object') to indexes <= 6
# Matching previous format nba_franchise_df.loc[nba_franchise_df.index >= 9, 'Average NBA Franchise Value (USD Millions)']
nba_rev_df.loc[nba_rev_df.index >= 6, 'NBA Sponsorship Revenue (USD Millions)'] =nba_rev_df.loc[nba_rev_df.index >= 6, 'NBA Sponsorship Revenue (USD Millions)'].apply(lambda x: '.' + x)

# Review efficacy
nba_rev_df
```

```python
# Round revenue column to 2 decimal places
nba_rev_df['NBA Sponsorship Revenue (USD Millions)'] = nba_rev_df['NBA Sponsorship Revenue (USD Millions)'].astype('float64').round(2)

# Check data type conversion & rounding in DF
nba_rev_df.info()
nba_rev_df
```

```python
# Change Column name
nba_rev_df = nba_rev_df.rename(columns={'NBA Sponsorship Revenue (USD Millions)':'NBA Sponsorship Revenue (USD Billions)'})

nba_rev_df.head(2)
```
