### NFL Revenue Data

```python
# Load in NFL team & league revenue csv
nfl_revenue_file = '../data/raw/MVP2/nfl_league_team_revenue.csv'
nfl_revenue_data = pd.read_csv(nfl_revenue_file)

# Create DataFrame
nfl_revenue_df = pd.DataFrame(nfl_revenue_data)
nfl_revenue_df.head()
```

```python
# Analyze DF characteristics & info
nfl_revenue_df.info()
```

**Cleaning Notes ^**
- Revenue column:
    - commas will need to changed to decimals 
    - Revenue numbers will need to be rounded to 1 decimal place to match billions format
    - Revenue column data type will need to be changed to float (decimal number)
    - Finally, rename column to billions for value

```python
# Convert data type of revenue column to float
nfl_revenue_df['NFL Revenue (USD Millions)'] = nfl_revenue_df['NFL Revenue (USD Millions)'].str.replace(',', '.', regex=False).astype(float)

# Check success of replace
nfl_revenue_df.head()
```

```python
# Round Revenue column to 1 decimal place
nfl_revenue_df['NFL Revenue (USD Millions)'] = nfl_revenue_df['NFL Revenue (USD Millions)'].round(1)

# Check rounding
nfl_revenue_df.head()
```

```python
# Rename revenue column 
nfl_revenue_df = nfl_revenue_df.rename(columns={'NFL Revenue (USD Millions)':'NFL Revenue ($USD billions)'})

# Verify change
nfl_revenue_df.head()
```
